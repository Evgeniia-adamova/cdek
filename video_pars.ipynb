# FACE RECOGNITION

```python
import cv2
import os

VIDEO_PATH = "video1.webm"

cap = cv2.VideoCapture(VIDEO_PATH)

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

frame_count = 0
face_count_total = 0

while frame_count < 50:
    ret, frame = cap.read()
    
    if not ret:
        break
        
    frame_count += 1
    
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    faces = face_cascade.detectMultiScale(
        gray, 
        scaleFactor=1.1, 
        minNeighbors=5, 
        minSize=(30, 30)
    )
    
    if len(faces) > 0:
        face_count_total += len(faces)
        print(f"Кадр {frame_count}: найдено {len(faces)} лиц")
        
        for (x, y, w, h) in faces:
            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        
        cv2.imshow('Видео с обнаруженными лицами', frame)
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()

print(f"\nОбработано кадров: {frame_count}")
print(f"Общее количество обнаруженных лиц: {face_count_total}")
```

```python
import cv2
import os
import numpy as np
from collections import Counter
import matplotlib.pyplot as plt

VIDEO_PATH = "video2.webm"

cap = cv2.VideoCapture(VIDEO_PATH)

frames_to_extract = [0, 30, 60, 90, 120]  # первые 5 кадров с шагом 30

for i, frame_num in enumerate(frames_to_extract):
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_num)
    ret, frame = cap.read()
    
    if ret:
        filename = f"extracted_frame_{i}.jpg"
        cv2.imwrite(filename, frame)
        print(f"Сохранен кадр {frame_num} как {filename}")
        
        plt.figure(figsize=(8, 6))
        plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        plt.title(f"Кадр {frame_num}")
        plt.axis('off')
        plt.show()

cap.release()
```

```python
frame_files = [f for f in os.listdir('.') if f.startswith('extracted_frame_') and f.endswith('.jpg')]
frame_files.sort() 

print(f"Найдено файлов кадров: {len(frame_files)}")
for f in frame_files:
    print(f"  - {f}")

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')


def detect_faces_in_image(image_path, scale_factor=1.1, min_neighbors=5, min_size=(30, 30)):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Не удалось загрузить изображение: {image_path}")
        return None, None
    
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=scale_factor,
        minNeighbors=min_neighbors,
        minSize=min_size
    )
    
    print(f"  Найдено лиц: {len(faces)}")
    
    return img, faces

for frame_file in frame_files:
    print(f"\nОбработка: {frame_file}")
    
    img, faces = detect_faces_in_image(frame_file)
    
    if img is None:
        continue
    
    img_with_faces = img.copy()
    
    for (x, y, w, h) in faces:
        cv2.rectangle(img_with_faces, (x, y), (x+w, y+h), (0, 255, 0), 2)
    
    output_filename = frame_file.replace('.jpg', '_faces.jpg')
    cv2.imwrite(output_filename, img_with_faces)
    print(f"  Сохранен результат: {output_filename}")
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    axes[0].set_title(f"Оригинал: {frame_file}")
    axes[0].axis('off')
    
    axes[1].imshow(cv2.cvtColor(img_with_faces, cv2.COLOR_BGR2RGB))
    axes[1].set_title(f"Обнаружено лиц: {len(faces)}")
    axes[1].axis('off')
    
    plt.tight_layout()
    plt.show()
```

```python
print("\n=== РАСШИРЕННОЕ ОБНАРУЖЕНИЕ ЛИЦ (DNN МОДЕЛЬ) ===")

# Попробуем использовать более современную DNN модель
def detect_faces_dnn(image_path):
    """
    Обнаружение лиц с использованием DNN модели
    """
    # Загружаем предварительно обученную модель
    # Скачиваем файлы модели если их нет
    model_files = {
        'config': 'deploy.prototxt',
        'model': 'res10_300x300_ssd_iter_140000.caffemodel'
    }
    
    # Проверяем наличие файлов модели
    need_download = False
    for file_name in model_files.values():
        if not os.path.exists(file_name):
            print(f"Файл модели не найден: {file_name}")
            need_download = True
    
    if need_download:
        print("Скачиваем модель DNN для обнаружения лиц...")
        import urllib.request
        
        # URL для файлов модели
        model_urls = {
            'config': 'https://raw.githubusercontent.com/opencv/opencv/master/samples/dnn/face_detector/deploy.prototxt',
            'model': 'https://raw.githubusercontent.com/opencv/opencv_3rdparty/dnn_samples_face_detector_20180205_fp16/res10_300x300_ssd_iter_140000_fp16.caffemodel'
        }
        
        for key, url in model_urls.items():
            print(f"Скачиваем {key}...")
            urllib.request.urlretrieve(url, model_files[key])
    
    # Загружаем модель
    net = cv2.dnn.readNetFromCaffe(model_files['config'], model_files['model'])
    
    # Загружаем изображение
    img = cv2.imread(image_path)
    if img is None:
        return None, []
    
    (h, w) = img.shape[:2]
    
    # Создаем blob из изображения
    blob = cv2.dnn.blobFromImage(cv2.resize(img, (300, 300)), 1.0,
                                 (300, 300), (104.0, 177.0, 123.0))
    
    # Подаем blob в сеть и получаем детекции
    net.setInput(blob)
    detections = net.forward()
    
    faces = []
    
    # Обрабатываем детекции
    for i in range(0, detections.shape[2]):
        confidence = detections[0, 0, i, 2]
        
        # Фильтруем слабые детекции
        if confidence > 0.5:
            # Вычисляем координаты ограничивающего прямоугольника
            box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])
            (startX, startY, endX, endY) = box.astype("int")
            
            faces.append((startX, startY, endX-startX, endY-startY, confidence))
    
    return img, faces

# Тестируем DNN модель на первом кадре
if frame_files:
    test_image = frame_files[0]
    print(f"Тестируем DNN модель на: {test_image}")
    
    img_dnn, faces_dnn = detect_faces_dnn(test_image)
    
    if img_dnn is not None:
        # Рисуем результаты
        img_result_dnn = img_dnn.copy()
        
        for (x, y, w, h, confidence) in faces_dnn:
            cv2.rectangle(img_result_dnn, (x, y), (x+w, y+h), (0, 255, 0), 2)
            text = f"{confidence:.2f}"
            cv2.putText(img_result_dnn, text, (x, y-10),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
        
        # Сравниваем результаты с Haar каскадом
        img_haar, faces_haar = detect_faces_in_image(test_image)
        img_result_haar = img_haar.copy()
        
        for (x, y, w, h) in faces_haar:
            cv2.rectangle(img_result_haar, (x, y), (x+w, y+h), (0, 255, 0), 2)
        
        # Показываем сравнение
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        
        axes[0].imshow(cv2.cvtColor(img_dnn, cv2.COLOR_BGR2RGB))
        axes[0].set_title("Оригинал")
        axes[0].axis('off')
        
        axes[1].imshow(cv2.cvtColor(img_result_haar, cv2.COLOR_BGR2RGB))
        axes[1].set_title(f"Haar каскад\nЛиц: {len(faces_haar)}")
        axes[1].axis('off')
        
        axes[2].imshow(cv2.cvtColor(img_result_dnn, cv2.COLOR_BGR2RGB))
        axes[2].set_title(f"DNN модель\nЛиц: {len(faces_dnn)}")
        axes[2].axis('off')
        
        plt.tight_layout()
        plt.show()
        
        print(f"\nСравнение методов:")
        print(f"  Haar каскад: {len(faces_haar)} лиц")
        print(f"  DNN модель: {len(faces_dnn)} лиц")
```

```python
print("\n=== ПАКЕТНАЯ ОБРАБОТКА ВСЕХ КАДРОВ ===")

# Создаем папку для результатов
os.makedirs('detected_faces', exist_ok=True)

# Выбираем метод (Haar или DNN)
USE_DNN = True  # Меняйте на False для использования Haar каскада

results = []

for frame_file in frame_files:
    print(f"\nОбработка: {frame_file}")
    
    if USE_DNN:
        img, faces = detect_faces_dnn(frame_file)
        face_count = len(faces)
    else:
        img, faces = detect_faces_in_image(frame_file)
        face_count = len(faces)
        faces = [(x, y, w, h, 1.0) for (x, y, w, h) in faces]  # Добавляем confidence=1.0
    
    if img is None:
        continue
    
    # Создаем результат с рамками
    img_result = img.copy()
    
    for i, (x, y, w, h, confidence) in enumerate(faces):
        # Рисуем прямоугольник
        color = (0, 255, 0)  # Зеленый
        cv2.rectangle(img_result, (x, y), (x+w, y+h), color, 2)
        
        # Добавляем текст с номером лица и уверенностью
        label = f"Face {i+1}"
        if confidence < 1.0:
            label += f" ({confidence:.2f})"
        
        cv2.putText(img_result, label, (x, y-10),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)
        
        # Сохраняем область лица как отдельное изображение
        face_roi = img[y:y+h, x:x+w]
        face_filename = f"detected_faces/{frame_file.replace('.jpg', '')}_face_{i+1}.jpg"
        cv2.imwrite(face_filename, face_roi)
    
    # Сохраняем полный результат
    result_filename = f"detected_faces/{frame_file.replace('.jpg', '')}_result.jpg"
    cv2.imwrite(result_filename, img_result)
    
    # Добавляем в результаты
    results.append({
        'filename': frame_file,
        'total_faces': face_count,
        'faces_details': [(x, y, w, h, conf) for (x, y, w, h, conf) in faces],
        'result_image': result_filename
    })
    
    print(f"  Найдено лиц: {face_count}")
    print(f"  Результат сохранен: {result_filename}")
    
    # Показываем результат
    plt.figure(figsize=(10, 6))
    plt.imshow(cv2.cvtColor(img_result, cv2.COLOR_BGR2RGB))
    plt.title(f"{frame_file} - Найдено лиц: {face_count}")
    plt.axis('off')
    plt.show()

print(f"\n=== ИТОГИ ===")
total_faces = sum(r['total_faces'] for r in results)
print(f"Всего обработано кадров: {len(results)}")
print(f"Всего обнаружено лиц: {total_faces}")
print(f"Среднее количество лиц на кадр: {total_faces/len(results):.1f}")

# Сохраняем статистику в файл
with open('detected_faces/summary.txt', 'w', encoding='utf-8') as f:
    f.write("=== СТАТИСТИКА ОБНАРУЖЕНИЯ ЛИЦ ===\n\n")
    f.write(f"Метод: {'DNN' if USE_DNN else 'Haar каскад'}\n")
    f.write(f"Всего обработано кадров: {len(results)}\n")
    f.write(f"Всего обнаружено лиц: {total_faces}\n\n")
    
    for i, result in enumerate(results):
        f.write(f"Кадр {i+1}: {result['filename']}\n")
        f.write(f"  Количество лиц: {result['total_faces']}\n")
        for j, (x, y, w, h, conf) in enumerate(result['faces_details']):
            f.write(f"    Лицо {j+1}: x={x}, y={y}, w={w}, h={h}")
            if conf < 1.0:
                f.write(f", confidence={conf:.3f}")
            f.write("\n")
        f.write("\n")

print("Статистика сохранена в: detected_faces/summary.txt")
```

```python
print("\n=== ВИЗУАЛИЗАЦИЯ СТАТИСТИКИ ===")

# Создаем график распределения лиц по кадрам
frame_numbers = list(range(1, len(results) + 1))
face_counts = [r['total_faces'] for r in results]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# График количества лиц
axes[0].bar(frame_numbers, face_counts, color='skyblue')
axes[0].set_xlabel('Номер кадра')
axes[0].set_ylabel('Количество лиц')
axes[0].set_title('Количество лиц на каждом кадре')
axes[0].set_xticks(frame_numbers)

# Добавляем значения на столбцы
for i, count in enumerate(face_counts):
    axes[0].text(i + 1, count + 0.1, str(count), ha='center', va='bottom')

# Pie chart распределения
if total_faces > 0:
    # Группируем по количеству лиц на кадре
    unique_counts = {}
    for count in face_counts:
        unique_counts[count] = unique_counts.get(count, 0) + 1
    
    labels = [f'{count} лиц' for count in unique_counts.keys()]
    sizes = list(unique_counts.values())
    
    axes[1].pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
    axes[1].set_title('Распределение кадров по количеству лиц')

plt.tight_layout()
plt.show()

# Выводим список сохраненных лиц
print("\n=== СОХРАНЕННЫЕ ИЗОБРАЖЕНИЯ ЛИЦ ===")
face_files = [f for f in os.listdir('detected_faces') if 'face_' in f and f.endswith('.jpg')]
print(f"Всего сохранено изображений лиц: {len(face_files)}")

if face_files:
    # Показываем первые 9 лиц
    n_to_show = min(9, len(face_files))
    fig, axes = plt.subplots(3, 3, figsize=(10, 10))
    
    for i in range(n_to_show):
        face_img = cv2.imread(f'detected_faces/{face_files[i]}')
        row, col = i // 3, i % 3
        
        axes[row, col].imshow(cv2.cvtColor(face_img, cv2.COLOR_BGR2RGB))
        axes[row, col].set_title(f"Лицо {i+1}")
        axes[row, col].axis('off')
    
    # Скрываем пустые subplots
    for i in range(n_to_show, 9):
        row, col = i // 3, i % 3
        axes[row, col].axis('off')
    
    plt.suptitle(f'Первые {n_to_show} обнаруженных лиц', fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = total_frames / fps
print(f"{fps}")
print(f"{total_frames}")
print(f"{duration}")
```

```python
frame_skip = 30  # Process every 30th frame
results = []
face_count = 0

for frame_idx in range(0, total_frames, frame_skip):
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
    ret, frame = cap.read()
    
    if not ret:
        continue
    
    # Convert to RGB (DeepFace expects RGB)
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    try:
        # Try multiple face detection backends
        detection_backends = ['opencv', 'ssd', 'mtcnn']
        
        for backend in detection_backends:
            try:
                # First, try to extract faces
                faces = DeepFace.extract_faces(
                    frame_rgb,
                    detector_backend=backend,
                    enforce_detection=False  # Don't fail if no faces
                )
                
                if faces and len(faces) > 0:
                    # Process each detected face
                    for i, face_info in enumerate(faces):
                        try:
                            # Analyze emotions for this face
                            analysis = DeepFace.analyze(
                                face_info['face'],
                                actions=['emotion'],
                                enforce_detection=False,
                                silent=True
                            )
                            
                            # Get dominant emotion
                            if isinstance(analysis, list):
                                analysis = analysis[0]
                            
                            emotions = analysis['emotion']
                            dominant_emotion = max(emotions.items(), key=lambda x: x[1])[0]
                            
                            results.append({
                                'frame': frame_idx,
                                'face_id': i,
                                'emotion': dominant_emotion,
                                'confidence': emotions[dominant_emotion],
                                'all_emotions': emotions,
                                'backend': backend
                            })
                            
                            face_count += 1
                            print(f"Frame {frame_idx}: Detected face {i+1} - {dominant_emotion} ({emotions[dominant_emotion]:.1f}%)")
                            
                        except Exception as e:
                            continue
                    
                    # Break if we found faces with this backend
                    if len(faces) > 0:
                        break
                        
            except Exception as e:
                continue
                
    except Exception as e:
        continue
    
    # Show progress
    if frame_idx % 300 == 0:
        print(f"Progress: {frame_idx}/{total_frames} frames")

cap.release()

print(f"\n✅ Processing complete!")
print(f"Total face detections: {face_count}")
```
